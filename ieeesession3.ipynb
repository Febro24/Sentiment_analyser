# SENTIMENT ANALYSER

*scikit-learn(SKLEARN) : a popular Python library for machine learning provides tools for classification, regression, clustering, and data preprocessing*

```python
import pandas as pd  #handling excel or csv type dataset
from sklearn.feature_extraction.text import CountVectorizer #COUNTVECTORIZER:text to numbers for machine understanding
from sklearn.naive_bayes import MultinomialNB #MultinomialNB = a text classifier that predicts labels based on word counts
```

SMALL HARD CODED LABELLED DATASET W/ INPUTS(TEXT) & OUTPUTS(EMOJIS)


```python
data = {
    "text": [
        "I am very happy today",
        "This is the best day ever",
        "I feel amazing",
        "I am feeling sad",
        "This is very disappointing",
        "I feel lonely and tired",
        "I am very angry right now",
        "This makes me furious",
        "I hate this so much"
    ],
    "emoji": ["😄","😄","😄","😢","😢","😢","😠","😠","😠"]
}


```

***dataframe : table with rows and columns***

```python
df = pd.DataFrame(data)
```

```python
df
```

**X:I AM VERY HAPPY [0, 1, 0, 1, 0, 1, 0] inputs in form of vector
Y:😄 output**

```python
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(df['text'])
y = df['emoji']
```

**Your data is text (sentences) and emojis.

Text is converted into numbers using CountVectorizer or TfidfVectorizer

MultinomialNB looks at these numbers and learns patterns like:

Words like "happy", "amazing", "best" → 😄

Words like "sad", "lonely", "tired" → 😢**

```python
model = MultinomialNB() #naive bayes classifier model assigning labels to inputs based on probabilities
model.fit(X, y) #fit: teaches the model the relationship between words (X) and emojis (y).
```
